```python
import csv
import re
import json
from datetime import datetime
from google.colab import files

# Upload file
uploaded = files.upload()
filename = list(uploaded.keys()) [0]

pattern = re.compile(r"^https://www\.medion\.com/(at|de)/shop(/.*)?$")

def extract(url):
  match = pattern.match(url.strip())
  if not match:
      raise ValueError(f"Invalid URL format: {url}")

  locale = match.group(1)
  path = match.group(2) or "/"
  return locale, path

redirects = []
seen = set()

with open(filename, newline='', encoding='utf-8') as csvfile:
  #Read the header row first to clean fieldnames
  raw_reader = csv.reader(csvfile, delimiter=';')
  header = next(raw_reader)
  # Clean the BOM character from fieldnames
  fieldnames = [h.replace('\ufeff', '') for h in header]

  #Create the DictReader with cleaned fieldnames. The raw_reader already
  #advanced teh file pointer, so DictReader will start reading from the
  #first data row.
  reader = csv.DictReader(csvfile, delimiter=';', fieldnames=fieldnames)

  for row in reader:
    from_url = row['old_url'].replace('\ufeff', '')
    to_url = row['new_url'].replace('\ufeff', '')

    from_locale, from_path = extract(from_url)
    to_locale, to_path = extract(to_url)

    # sanity checks
    if from_locale != to_locale:
        raise ValueError(f"Locale mismatch: {from_url} -> {to_url}")

    key = f"{from_locale}:{from_path}"
    if key in seen:
        raise ValueError(f"Duplicate redirect: {key}")
    seen.add(key)

    if from_path == to_path:
        raise ValueError(f"Self redirect: {from_url}")

    redirects.append({
        "locale": from_locale,
        "from": from_path,
        "to": to_path,
        "permanent": True
    })

  # Output JSON
  print(json.dumps(redirects, indent=2, ensure_ascii=False))
  output_filename = f"{datetime.now().strftime('%Y%m%d')}_redirects.json"

  with open(output_filename, "w", encoding="utf-8") as f:
    json.dump(redirects, f, indent=2, ensure_ascii=False)

  files.download(output_filename)


```
